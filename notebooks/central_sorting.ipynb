# Central Z-Lab sorting

One configuration controls loading, probe wiring, sorting/clustering, curation and saved inspection. Start with an example TOML file, edit its paths and verify wiring. [API guide](../docs/api.md). No cell has been executed.


```python
from pathlib import Path
from sorting_scripts import validate, run, status

# Copy/edit an examples/*.toml template, then set its absolute path here.
# Input/output paths inside TOML resolve relative to that TOML file.
CONFIG = Path("/absolute/path/to/session.toml").expanduser().resolve()
```

## Validate before computing

This opens headers and checks groups without sorting, caching, downloading models or computing extensions. Review channel IDs and groups.


```python
validation = validate(CONFIG)
validation
```

## Run

Creates a unique output directory. Models default off. `labels_only` preserves units/events; other curation policies are explicit in the configuration.


```python
RUN_DIR = run(CONFIG)
manifest = status(RUN_DIR)
manifest["outputs"]
```

## Inspect stored results

This reads existing waveforms and metrics. It never computes missing extensions. An empty sorting has no final analyzer.


```python
from sorting_scripts.inspection import open_analyzer, summary_figure

if manifest["outputs"]["analyzer"] is not None:
    analyzer = open_analyzer(RUN_DIR / manifest["outputs"]["analyzer"])
    UNIT = analyzer.unit_ids[0]
    figure, waveforms, metrics = summary_figure(analyzer, UNIT)
    display(figure)
    display(metrics)
```

## Resume and interactive curation

Resume an interrupted/failed run with `run(resume="/absolute/run/path")`. Completed group artifacts and input/code/environment identities are checked. Failed attempts are preserved. For interactive review, explicitly call `from sorting_scripts.gui import launch_curation_gui; launch_curation_gui(analyzer)`. Stored-only inspection and interactive curation are separate operations.
